```python
def sequential_search(arr, x):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1


data = [10, 50, 30, 70, 80, 20]
index = sequential_search(data, 30)

if index != -1:
    print(f"Indeks: {index}")
else:
    print("Data tidak ditemukan.")
```

```python
def binary_search(arr, x):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return -1


data = [2, 5, 8, 12, 16, 23, 38]
index = binary_search(data, 23)

if index != -1:
    print(f"Indeks: {index}")
else:
    print("Data tidak ditemukan.")
```

```python
def find_element_linear(table, z):
    for i in range(len(table)):
        if table[i]["z"] == z:
            return i
    return -1


def find_element_binary(table, z):
    low, high = 0, len(table) - 1
    while low <= high:
        mid = low + (high - low) // 2
        if table[mid]["z"] == z:
            return mid
        elif table[mid]["z"] < z:
            low = mid + 1
        else:
            high = mid - 1
    return -1


table = [
    {"z": 1, "density": 0.000089, "name": "Hidrogen"},
    {"z": 2, "density": 0.000178, "name": "Helium"},
    {"z": 6, "density": 2.26, "name": "Karbon"},
    {"z": 13, "density": 2.70, "name": "Aluminium"},
    {"z": 14, "density": 2.33, "name": "Silikon"},
    {"z": 26, "density": 7.87, "name": "Besi"},
    {"z": 29, "density": 8.96, "name": "Tembaga"},
    {"z": 79, "density": 19.32, "name": "Emas"}
]

try:
    z = int(input("Nomor atom: "))
except ValueError:
    raise SystemExit("Masukan harus berupa bilangan bulat.")
if z <= 0:
    raise SystemExit("Nomor atom harus positif.")

i = find_element_binary(table, z)
j = find_element_linear(table, z)
assert i == j

if i == -1:
    print("Unsur tidak terdapat dalam tabel parsial.")
else:
    el = table[i]
    print(f"{el['name']} (Z={el['z']}): {el['density']:.6g} g/cm^3")
```

```python
def bubble_sort(a):
    n = len(a)
    for i in range(n - 1):
        changed = False
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                changed = True
        if not changed:
            break
```

```python
def selection_sort(a):
    for i in range(len(a) - 1):
        m = i
        for j in range(i + 1, len(a)):
            if a[j] < a[m]:
                m = j
        if m != i:
            a[i], a[m] = a[m], a[i]
```

```python
def insertion_sort(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
```

```python
def merge(a, tmp, l, m, r):
    i, j, k = l, m, l
    while i < m and j < r:
        if a[i] <= a[j]:
            tmp[k] = a[i]
            i += 1
        else:
            tmp[k] = a[j]
            j += 1
        k += 1
    while i < m:
        tmp[k] = a[i]
        i += 1
        k += 1
    while j < r:
        tmp[k] = a[j]
        j += 1
        k += 1
    a[l:r] = tmp[l:r]


def merge_sort(a, tmp, l, r):
    if r - l <= 1:
        return
    m = l + (r - l) // 2
    merge_sort(a, tmp, l, m)
    merge_sort(a, tmp, m, r)
    merge(a, tmp, l, m, r)
```

```python
def partition(a, low, high):
    pivot = a[high]
    i = low
    for j in range(low, high):
        if a[j] < pivot:
            a[i], a[j] = a[j], a[i]
            i += 1
    a[i], a[high] = a[high], a[i]
    return i


def quick_sort(a, low, high):
    if low >= high:
        return
    p = partition(a, low, high)
    quick_sort(a, low, p - 1)
    quick_sort(a, p + 1, high)
```
